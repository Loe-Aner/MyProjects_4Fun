Dostajesz listę zdarzeń walki — każde to krotka (gracz, obrażenia). Boss zawsze atakuje gracza z najwyższym łącznym threatem (sumą zadanych obrażeń). Po każdym zdarzeniu zwróć, kogo boss aktualnie atakuje. Przy remisie boss nie zmienia celu — trzyma się tego, kogo bił do tej pory.

```python
def aggro(events: list[tuple[str, int]]) -> list[str]:
    """
    Tłucze gracza który zadaje najwięcej dps.
    """
    current_target = None
    gracze = []
    dps_total = {gracz: 0 for gracz, _ in events}
    for gracz, threat in events:
        dps_total[gracz] += threat

        if current_target is None or dps_total[gracz] > dps_total[current_target]:
            current_target = gracz
        gracze.append(current_target)
        
    return gracze
```

```python
aggro([("Jaina", 50), ("LoeAner", 10), ("Thrall", 100), ("Jaina", 50), ("LoeAner", 200)])
```
