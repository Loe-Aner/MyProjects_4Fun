Dostajesz missing — słownik {gracz: brakujące_HP} — oraz listę leczeń (gracz, ilość). Każde leczenie uzupełnia brakujące HP danego gracza; wszystko ponad jego brak to overheal i przepada. Zwróć łączne efektywne leczenie (bez overhealu).

```python
from itertools import groupby
from operator import itemgetter
```

```python
def effective_healing(missing: dict[str, int], heals: list[tuple[str, int]]) -> int:
    remaining = dict(missing)
    total_effective = 0

    for player, heal in heals:
        missing_hp = remaining.get(player, 0)
        effective = min(missing_hp, heal)
        total_effective += effective
        remaining[player] = missing_hp - effective
        
    return total_effective
```

```python
effective_healing(
    {"Thrall": 100, "Jaina": 50},
    [("Thrall", 60), ("Jaina", 80), ("Thrall", 60)],
)
```
