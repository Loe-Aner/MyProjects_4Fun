Masz log wejść graczy do instancji (poniżej), gdzie każdy wpis to (gracz, rola). Twoim zadaniem jest pogrupować graczy po roli i wypisać listę nicków w każdej roli — pamiętając o pułapce z sąsiedztwem. Następnie, dla jednego punktu ekstra, policz liczność każdej roli nie używając Counter (tym razem ma być przez groupby, żeby poczuć różnicę między narzędziami). Zwróć uwagę, co musisz zrobić zanim odpalisz groupby.

```python
party = [
    ("Thrall", "DPS"),
    ("Jaina", "Healer"),
    ("Sylvanas", "DPS"),
    ("Uther", "Tank"),
    ("Rexxar", "DPS"),
    ("Anduin", "Healer"),
]
```

```python
from operator import itemgetter
from itertools import groupby
from collections import defaultdict
```

```python
sorted(party, key=lambda n: n[1])
```

```python
s = defaultdict(list)
for gracz, rola in sorted(party, key=itemgetter(1)):
    s[rola].append(gracz)
s
```

```python
key = itemgetter(1)
```

```python
for rola, grupa in groupby(sorted(party, key=key), key=key):
    print(rola, [p[0] for p in grupa])
```

```python
for rola, grupa in groupby(sorted(party, key=key), key=key):
    print(rola, sum(1 for _ in grupa))
```
