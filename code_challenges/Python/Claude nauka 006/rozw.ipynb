Masz log itemlevelu gracza z kolejnych tygodni rajdowania (poniżej). Trzy rzeczy: (1) używając pairwise, policz przyrost ilvl tydzień do tygodnia; (2) używając accumulate z max, pokaż „rekordowy ilvl do danego tygodnia" — czyli najwyższy osiągnięty do tej pory; (3) pewnego tygodnia gracz miał spadek (sprzedał zły item) — znajdź, w którym tygodniu nastąpił pierwszy spadek względem poprzedniego. Spróbuj bez ręcznego indeksowania range(len(...)).

```python
weekly_ilvl = [340, 350, 350, 372, 368, 385]
```

```python
from itertools import pairwise, accumulate
```

```python
[(b - a) for a, b in pairwise(weekly_ilvl)]
```

```python
list(accumulate(weekly_ilvl, max))
```
