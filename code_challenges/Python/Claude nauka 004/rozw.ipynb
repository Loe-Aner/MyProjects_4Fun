Masz płaską listę par (boss, item) z całego raidu — ten sam boss występuje wielokrotnie. Zbuduj słownik boss → lista jego dropów, używając defaultdict. Następnie dorzuć jeden warunek: pomiń duplikaty, żeby każdy przedmiot pod danym bossem pojawił się tylko raz (podpowiedź: jest inna fabryka niż list, która robi to za darmo). Bez ani jednego if key in dict.

```python
raid_log = [
    ("Ragnaros", "Sulfuras"),
    ("Garr", "Aurastone Hammer"),
    ("Ragnaros", "Sulfuras"),
    ("Ragnaros", "Band of Accuria"),
    ("Garr", "Brutality Blade"),
]
```

```python
from collections import defaultdict
```

```python
d = defaultdict(set)
for boss, item in raid_log:
    d[boss].add(item)

d
```
